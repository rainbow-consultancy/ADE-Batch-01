```python
container_name = "databricks-container"
storage_account_name = "rainbowstorageadls"
client_id = dbutils.secrets.get(scope="sc-rainbow-vault", key="clinet-id")
tenant_id = dbutils.secrets.get(scope="sc-rainbow-vault", key="tenent-id")
client_secret = dbutils.secrets.get(scope="sc-rainbow-vault", key="client-secret")
mount_path = "/mnt/rainbowcontainer"
```

```python
configs = {
    "fs.azure.account.auth.type": "OAuth",
    "fs.azure.account.oauth.provider.type": "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider",
    "fs.azure.account.oauth2.client.id": client_id,
    "fs.azure.account.oauth2.client.secret": client_secret,
    "fs.azure.account.oauth2.client.endpoint": f"https://login.microsoftonline.com/{tenant_id}/oauth2/token"
}

# Optionally, you can add <directory-name> to the source URI of your mount point.
dbutils.fs.mount(
  source = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/",
  mount_point = mount_path,
  extra_configs = configs
)
```
