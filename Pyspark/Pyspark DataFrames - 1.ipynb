```python
from pyspark.sql import functions as F
from pyspark.sql.types import IntegerType, StringType, StructField, StructType
```

```python
# 1. WithOut Strict Schema

row_data = [
    (100, 'Sandeep', 30000),
    (101, 'Gagan', 20000),
    (102, 'Kiran', 35000),
    (103, 'Krishna', 15000)
]

columns = ["EmpId", "EmpName", "Salary"]

emp_df = spark.createDataFrame(data=row_data, schema=columns)
display(emp_df)
```

```python
emp_df.show()
```

```python
# 2. With Strict Schema

row_data = [
    (100, 'Sandeep', 30000),
    (101, 'Gagan', 20000),
    (102, 'Kiran', 35000),
    (103, 'Krishna', 15000)
]

columns = StructType(
    [
        StructField("EmpId", IntegerType(), False),
        StructField("EmpName", StringType(), True),
        StructField("Salary", IntegerType(), True)
    ]
)

emp_df = spark.createDataFrame(data=row_data, schema=columns)
display(emp_df)
```

Add a new column into bihar_df called Country and also add default value "India" to that column

```python
# withColumn

bihar_df = bihar_df.withColumn('Country', F.lit("India"))
bihar_df.display()
```

```python
df = (
    bihar_df
    .withColumn('Country', F.lit("India"))
    .withColumn('Test1', F.lit("Value1"))
    .withColumn('Test2', F.lit("Value2"))
    .withColumn('Test3', F.lit("Value3"))
)
df.display()
```

Rename a column

```python
# withColumnRenamed

bihar_df = (
  bihar_df
  .withColumnRenamed("Constituency Number", "ConstituencyNumber")
  .withColumnRenamed("Constituency Name", "ConstituencyName")
  .withColumnRenamed("Serial Number", "SerialNumber")
  .withColumnRenamed("Candidate Name", "CandidateName")
  .withColumnRenamed("EVM Votes", "EvmVotes")
  .withColumnRenamed("Postal Votes", "PostalVotes")
  .withColumnRenamed("Total Votes", "TotalVotes")
  .withColumnRenamed("Vote Percentage", "VotePercentage")
)
bihar_df.display()
```
