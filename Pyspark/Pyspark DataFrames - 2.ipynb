```python
df = (
    bihar_df
    .select(
        F.col("Constituency Number").alias("ConstituencyNumber"), 
        F.col("Constituency Name").alias("ConstituencyName")
    )
)
df.display()
```

```python
df.distinct().display()
```

```python
# drop
df.drop("ConstituencyNumber").display()
```

```python
# dropDuplicates

df = bihar_df.dropDuplicates(subset=["Constituency Name", "Party"])
df.display()
```

```python
df = bihar_df.dropDuplicates()
df.display()
```

```python
df.orderBy(F.col("Constituency Name").asc()).display()
```

```python
df.orderBy(F.col("Constituency Number").asc()).display()
```

```python
df.orderBy(F.col("Constituency Number").desc()).display()
```

```python
df.sort(F.col("Constituency Number")).display()
```

```python
# group by

participation_df = (
    bihar_df
    .groupBy("Constituency Number", "Constituency Name")
    .count()
)

(
    participation_df
    .orderBy(F.col("Constituency Number").asc())
    .display()
)
```

```python
# group by

participation_df = (
    bihar_df
    .groupBy("Constituency Number", "Constituency Name")
    .agg(
        F.count("*").alias("TotalParticipents")
    )
)

(
    participation_df
    .orderBy(F.col("Constituency Number").asc())
    .display()
)
```

```python
# group by

participation_df = (
    bihar_df
    .groupBy("Constituency Number", "Constituency Name")
    .agg(
        F.count("*").alias("TotalParticipents"),
        F.min("Postal Votes").alias("MinPostalVotes"),
        F.max("Postal Votes").alias("MaxPostalVotes")
    )
)

(
    participation_df
    .orderBy(F.col("Constituency Number").asc())
    .display()
)
```
