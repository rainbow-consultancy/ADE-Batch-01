```python
from pyspark.sql import functions as F
```

```python
loans_df = spark.read.format("parquet").load("/Volumes/workspace/default/kaggle_datasets/Loans")
loans_df.display()
```

```python
loans_df.groupBy("ACTIONTYPE").agg(F.count("ACTIONTYPE")).display()
```

```python
loans_df.count()
```

```python
loans_df.write.format("delta").mode("overwrite").save("/Volumes/workspace/default/kaggle_datasets/DeltaLoans")
```

```python
# loans_df_filtered = loans_df.filter(F.col("ACTIONTYPE") != "A")
loans_df_filtered = loans_df.filter((F.col("ACTIONTYPE") == "A"))
loans_df_filtered.count()
```

```python
loans_df_filtered.write.format("delta").mode("overwrite").save("/Volumes/workspace/default/kaggle_datasets/DeltaLoans")
```

```python
df = spark.read.format("delta").load("/Volumes/workspace/default/kaggle_datasets/DeltaLoans")
df.count()
```
