```python
spark.conf.set("spark.sql.adaptive.enabled", "false")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "-1")
```

```python
# joins
```

```python
from pyspark.sql import Row
```

```python
student_df = spark.createDataFrame([
    Row(stuid=1, name='Alice', depid=101),
    Row(stuid=2, name='Bob', depid=102),
    Row(stuid=3, name='Charlie', depid=101),
    Row(stuid=4, name='David', depid=103),
    Row(stuid=5, name='John', depid=105)
])

display(student_df)
```

```python
department_df = spark.createDataFrame([
    Row(depid=101, dep_name='Computer Science'),
    Row(depid=102, dep_name='Mathematics'),
    Row(depid=103, dep_name='Physics'),
    Row(depid=104, dep_name='Electronics'),
])

display(department_df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=["depid"],
        how="inner"  # --> inner join is optional
    )
)

display(df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=["depid"],
        how="left"
    )
)

display(df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=student_df["depid"] == department_df["depid"],
        how="right"
    )
    .select(student_df["*"], department_df["dep_name"])
)

display(df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=student_df["depid"] == department_df["depid"],
        how="full"
    )
    .select(student_df["*"], department_df["dep_name"])
)

display(df)
```

```python
# identify students who is not having their depid in the department table

df = (
    student_df
    .join(
        department_df,
        on=student_df["depid"] == department_df["depid"],
        how="leftanti"
    )
)

display(df)
```

```python
# identify students who is not having their depid in the department table

df = (
    student_df
    .join(
        department_df,
        on=student_df["depid"] == department_df["depid"],
        how="left"
    )
    .filter(department_df["depid"].isNull())
    .select(student_df["*"])
)

display(df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=student_df["depid"] == department_df["depid"],
        how="leftsemi"
    )
)

display(df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=student_df["depid"] == department_df["depid"],
        how="inner"
    )
    .select(student_df["*"])
)

display(df)
```

```python
# cross join  -- return cartesian product of the two tables

df = (
    student_df
    .crossJoin(department_df)
)

display(df)
```

```python
student_df.createOrReplaceTempView("student")
department_df.createOrReplaceTempView("department")
```
