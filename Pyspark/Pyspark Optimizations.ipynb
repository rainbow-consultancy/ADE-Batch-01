```python
spark.conf.set("spark.sql.adaptive.enabled", "true")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "104857600")
```

```python
from pyspark.sql import Row
from pyspark.sql import functions as F
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
student_df = spark.createDataFrame([
    Row(stuid=1, name='Alice', depid=101),
    Row(stuid=2, name='Bob', depid=102),
    Row(stuid=3, name='Charlie', depid=101),
    Row(stuid=4, name='David', depid=103),
    Row(stuid=5, name='John', depid=105)
])

display(student_df)
```

```python
department_df = spark.createDataFrame([
    Row(depid=101, dep_name='Computer Science'),
    Row(depid=102, dep_name='Mathematics'),
    Row(depid=103, dep_name='Physics'),
    Row(depid=104, dep_name='Electronics'),
])

display(department_df)
```

```python
df = (
    student_df
    .join(
        department_df,
        on=["depid"],
        how="inner"  # --> inner join is optional
    )
)

display(df)
```

```python
df = (
    student_df
    .join(
        F.broadcast(department_df),
        on=["depid"],
        how="inner"  # --> inner join is optional
    )
)

display(df)
```
