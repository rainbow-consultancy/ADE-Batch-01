```python
from pyspark.sql import functions as F, types as T
```

```python
dbutils.fs.mounts()
```

```python
# Read dataset from adls location
mount_path = "/mnt/rainbowcontainer"
file_path = "raw-datasets/bihar_election_results.csv"
bihar_df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{mount_path}/{file_path}")
bihar_df.display()
```

```python
bihar_df = bihar_df.withColumn("Constituency Name", F.lower(F.col("Constituency Name")))
bihar_df.display()
```

```python
# 1. method
def get_even_serial_num(val):
    return val % 2 == 0

# --> register the function as UDF
get_even_serial_num_udf = F.udf(get_even_serial_num, T.BooleanType())
```

```python
display(
    bihar_df
    .filter(
        get_even_serial_num_udf(F.col("Serial Number"))
    )
)
```

```python
get_even_serial_num(20)
```

```python
# 2. method
@udf(T.BooleanType())
def get_odd_serial_num(val):
    return val % 2 != 0
```

```python
display(
    bihar_df
    .filter(
        get_odd_serial_num(F.col("Serial Number"))
    )
)
```
